```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.cluster import KMeans

dataset=pd.read_csv('/content/Mall_Customers.csv')

x = dataset.iloc[:,[3,4]].values

wcss_list = []

for i in range(1,11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(x)
    wcss_list.append(kmeans.inertia_)

plt.plot(range(1,11),wcss_list)

plt.title('The elbow method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS_list')
plt.show()
```

```python
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42, n_init=10)
y_predict = kmeans.fit_predict(x)

plt.scatter(x[y_predict==0,0], x[y_predict==0,1], s=100, c='indigo', label='Cluster 1')
plt.scatter(x[y_predict==1,0], x[y_predict==1,1], s=100, c='gold', label='Cluster 2')
plt.scatter(x[y_predict==2,0], x[y_predict==2,1], s=100, c='red', label='Cluster 3')
plt.scatter(x[y_predict==3,0], x[y_predict==3,1], s=100, c='black', label='Cluster 4')
plt.scatter(x[y_predict==4,0], x[y_predict==4,1], s=100, c='brown', label='Cluster 5')
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], s=300, c='coral', label='Centroids')

plt.title('Clusters of customers')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.show()
```

## Interpretation of Customer Clusters

Based on the KMeans clustering of Annual Income and Spending Score, we can identify five distinct customer segments:

*   **Cluster 1 (Indigo):** These customers have **average annual income** and an **average spending score**. They represent a mid-range customer segment.
*   **Cluster 2 (Gold):** This segment consists of customers with **high annual income** but a **low spending score**. These might be cautious spenders or those who prefer luxury items less frequently.
*   **Cluster 3 (Red):** These are customers with **low annual income** but a **high spending score**. This group might include impulse buyers or those spending a larger proportion of their income.
*   **Cluster 4 (Black):** This cluster represents customers with **low annual income** and a **low spending score**. They are likely budget-conscious customers.
*   **Cluster 5 (Brown):** These customers have **high annual income** and a **high spending score**. They are often considered the target customers for premium products and services, exhibiting high purchasing power and willingness to spend.

This segmentation can be very useful for targeted marketing strategies, product development, and personalized customer experiences.
